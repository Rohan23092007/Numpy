```python
import numpy as np
```

```python
#Numpy Indexing and Slicing of Vector
```

```python
arr = np.arange(11,21)
arr
```

```python
arr[5]
```

```python
arr[1:5]
```

```python
arr[:]
```

```python
arr[3:]
```

```python
arr_slice = arr[3::2]
arr_slice
```

```python
#Numpy Indexing and Slicing of Matrix
```

```python
arr = np.arange(1,31).reshape(6,5)
arr
```

```python
arr[0]
```

```python
arr[4,3]
```

```python
arr[3:7,3:5] #3 start index of row: 7 end index of row goes till 6 , 3 start index of column : 5 end index of column goes till 4
```

```python
arr[:,2]
```

```python
#Boolean Indexing
```

```python
arr = np.arange(11,21)
arr
```

```python
bool_index = arr % 2 == 0
bool_index
```

```python
arr = arr[bool_index]
arr
```
